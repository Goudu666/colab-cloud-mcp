# Colab Cloud MCP Server (Cloud-to-Colab 直连启动器)

让 **Claude 云端对话（网页 / 手机 App / Desktop Custom Connectors）**、**Google Antigravity** 和 **Claude Code** 通过公网 HTTPS (443) `Streamable HTTP` 直接连接 Google Colab 运行时——**无需本机常开 CLI 或浏览器桥接，合上笔记本电脑也能继续在手机或云端指挥 GPU 跑实验。**

### 推荐一次性配置（Colab 左侧 Secrets 面板）
- `MCP_TOKEN`：自定义一串高强度密钥（固定后每次重启无需重配 Token）
- *(可选，固定域名免改 URL)*：
  - **Cloudflare Named Tunnel**：填入 `CLOUDFLARE_TUNNEL_TOKEN` 与 `CLOUDFLARE_HOSTNAME`（转发至本地端口 `8765`）
  - **或 ngrok 免费静态域名**：填入 `NGROK_AUTHTOKEN` 与 `NGROK_DOMAIN`

```python
# [可选] Step 1: 挂载 Google Drive，方便后台训练日志与模型权重持久化保存
MOUNT_GOOGLE_DRIVE = False  # 改为 True 即可自动挂载到 /content/drive

if MOUNT_GOOGLE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')
```

```python
# Step 2: 从 GitHub 安装 colab-cloud-mcp
import subprocess, sys
repo_url = 'git+https' + '://' + 'github' + '.com/Goudu666/colab-cloud-mcp.git'
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', repo_url])
```

```python
# Step 3: 启动 Remote MCP 服务与 HTTPS 隧道，打印 Claude / Antigravity 连接配置
from colab_cloud_mcp import start_colab_mcp

mcp_server = start_colab_mcp(
    provider="auto",
    port=8765,
    block=False,
)
```

```python
# Step 4: 前台保活监控（合上电脑前运行此单元格，使 Colab 保持活跃执行状态）
mcp_server.keep_alive(status_interval_seconds=60)
```
